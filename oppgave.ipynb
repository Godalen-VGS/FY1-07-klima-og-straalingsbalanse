Hvorfor har jorda en gjennomsnittstemperatur på omtrent 15 °C? Svaret er en **balanse**: jorda tar imot energi fra sola og stråler like mye energi ut igjen. Her lager du en enkel modell av denne balansen, og ser hvor stor forskjell drivhuseffekten gjør.

## Strålingslovene

**Stefan–Boltzmanns lov:** En overflate med temperatur $T$ (i kelvin) stråler ut effekten

$$\frac{P}{A} = \sigma T^4 \qquad \sigma = 5{,}67 \cdot 10^{-8}\ \text{W/(m}^2\text{K}^4)$$

per kvadratmeter.

**Wiens forskyvningslov:** Strålingen er sterkest ved bølgelengden

$$\lambda_{\text{maks}} = \frac{b}{T} \qquad b = 2{,}898 \cdot 10^{-3}\ \text{m·K}$$

## Energibalansen

- Sola leverer $S = 1361$ W/m² til en flate vinkelrett på solstrålene, rett utenfor atmosfæren (**solarkonstanten**).
- Jorda reflekterer andelen $\alpha \approx 0{,}30$ rett tilbake (**albedo**).
- Jorda fanger opp sollys på en sirkelflate $\pi R^2$, men stråler ut fra hele kuleoverflaten $4\pi R^2$. Derfor deler vi på 4.

Likevekt betyr at innkommende effekt er lik utstrålt effekt:

$$\frac{S(1 - \alpha)}{4} = \sigma T^4$$
